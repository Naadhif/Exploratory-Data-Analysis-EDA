```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)  
jumlah_data = 100

nim = [f"2024MAHA{i:03d}" for i in range(1, jumlah_data + 1)]

opsi_jurusan = ["Teknik Informatika", "Sistem Informasi", "Data Science", "Teknik Komputer"]
jurusan = np.random.choice(opsi_jurusan, size=jumlah_data)

opsi_kategori_kehadiran = ["Rendah", "Sedang", "Tinggi"]
kategori_kehadiran = np.random.choice(opsi_kategori_kehadiran, size=jumlah_data)

opsi_status_lulus = ["Lulus", "Tidak Lulus"]
status_lulus = np.random.choice(opsi_status_lulus, size=jumlah_data, p=[0.75, 0.25])

jam_belajar_per_minggu = np.random.randint(5, 45, size=jumlah_data)
persentase_kehadiran = np.round(np.random.uniform(50.0, 100.0, size=jumlah_data), 2)
skor_ujian_tengah = np.round(np.random.uniform(40.0, 100.0, size=jumlah_data), 2)
ipk = np.round(np.random.uniform(2.0, 4.0, size=jumlah_data), 2)

data_mahasiswa = {
    "NIM": nim,
    "Jurusan": jurusan,
    "Kategori_Kehadiran": kategori_kehadiran,
    "Status_Lulus": status_lulus,
    "Jam_Belajar_Minggu": jam_belajar_per_minggu,
    "Persentase_Kehadiran": persentase_kehadiran,
    "Skor_Ujian_Tengah": skor_ujian_tengah,
    "IPK": ipk
}

df = pd.DataFrame(data_mahasiswa)

display(df.head())

nama_file = "dataset_performa_mahasiswa_100.csv"
df.to_csv(nama_file, index=False)
```

```python

df = pd.read_csv('dataset_performa_mahasiswa_100.csv')
```

```python

print("=== HEAD (5 Data Pertama) ===")
display(df.head())

print("\n=== SHAPE (Ukuran Dataset) ===")
print("Baris, Kolom:", df.shape)

print("\n=== COLUMNS (Daftar Atribut) ===")
print(df.columns.tolist())

print("\n=== INFO (Informasi Struktur) ===")
df.info()

print("\n=== DTYPES (Tipe Data) ===")
print(df.dtypes)
```

```python

print("\n=== KLASIFIKASI JENIS ATRIBUT ===")
jenis_atribut = {
    "NIM": "Identifier",
    "Jurusan": "Kategorikal Nominal",
    "Kategori_Kehadiran": "Kategorikal Ordinal",
    "Status_Lulus": "Kategorikal Biner",
    "Jam_Belajar_Minggu": "Numerik Diskrit",
    "Persentase_Kehadiran": "Numerik Kontinu",
    "Skor_Ujian_Tengah": "Numerik Kontinu",
    "IPK": "Numerik Kontinu"
}
for kolom, jenis in jenis_atribut.items():
    print(f"- {kolom:<22} : {jenis}")
```

```python

fitur_numerik = ["Jam_Belajar_Minggu", "Persentase_Kehadiran", "Skor_Ujian_Tengah", "IPK"]
display(df[fitur_numerik].describe().round(2))
```

```python

fitur_kategorikal = ["Jurusan", "Kategori_Kehadiran", "Status_Lulus"]

for kolom in fitur_kategorikal:
    print(f"\n=== Atribut: {kolom} ===")
    frekuensi = df[kolom].value_counts()
    persentase = df[kolom].value_counts(normalize=True).mul(100).round(2)
    display(pd.DataFrame({"Frekuensi": frekuensi, "Persentase (%)": persentase}))
```

```python

print("===HISTOGRAM===")
fitur_hist = ["Jam_Belajar_Minggu", "IPK"]
for kolom in fitur_hist:
    plt.figure(figsize=(8, 5))
    sns.histplot(
        data=df,
        x=kolom,
        bins=10,
        kde=True
    )
    plt.title(f"Distribusi {kolom}")
    plt.xlabel(kolom)
    plt.ylabel("Frekuensi")
    plt.show()
```

```python

print("===COUNT PLOT===")
fitur_count = ["Jurusan", "Kategori_Kehadiran"]
for kolom in fitur_count:
    plt.figure(figsize=(7, 5))
    sns.countplot(
        data=df,
        x=kolom,
        hue=kolom,
        legend=False
    )
    plt.title(f"Distribusi {kolom}")
    plt.xlabel(kolom)
    plt.ylabel("Jumlah")
    plt.xticks(rotation=15 if kolom == "Jurusan" else 0)
    plt.show()
```

```python

print("===PIE CHART===")
status_counts = df["Status_Lulus"].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(
    status_counts.values,
    labels=status_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Proporsi Status Kelulusan Mahasiswa")
plt.show()
```

```python

print("===SCATTER PLOT===")
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="Jam_Belajar_Minggu",
    y="IPK"
)
plt.title("Hubungan Jam Belajar per Minggu dengan IPK")
plt.xlabel("Jam Belajar per Minggu")
plt.ylabel("IPK")
plt.show()

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="Skor_Ujian_Tengah",
    y="IPK"
)
plt.title("Hubungan Skor Ujian Tengah dengan IPK")
plt.xlabel("Skor Ujian Tengah")
plt.ylabel("IPK")
plt.show()
```
